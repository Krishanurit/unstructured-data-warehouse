```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from warehouse.connections import get_connection
```

```python
conn = get_connection()

df = pd.read_sql_query("SELECT * FROM fact_data", conn)
df.head()
```

```python
keyword_counts = df['keyword'].value_counts().head(10)

plt.figure(figsize=(8,5))
sns.barplot(x=keyword_counts.values, y=keyword_counts.index)
plt.title("Top Keywords")
plt.show()
```

```python
sentiment_counts = df['sentiment'].value_counts()

plt.figure(figsize=(5,5))
plt.pie(sentiment_counts, labels=sentiment_counts.index, autopct='%1.1f%%')
plt.title("Sentiment Distribution")
plt.show()
```

```python
query = """
SELECT c.category_name, COUNT(*) AS total
FROM fact_data f
JOIN dim_category c ON f.category_id = c.category_id
GROUP BY c.category_name;
"""

cat_df = pd.read_sql_query(query, conn)

sns.barplot(x='total', y='category_name', data=cat_df)
plt.title("Category Distribution")
plt.show()
```
